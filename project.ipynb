#D. Группировки и агрегации

```python
average_purchase_by_gender = df.groupby('Gender')['Purchase Amount (USD)'].mean()
print("Average Purchase Amount (USD) by Gender:")
print(average_purchase_by_gender)
```

```python
item_count_by_category = df.groupby('Category')['Item Purchased'].count()
print("Number of Item Purchased by Category:")
print(item_count_by_category)
```

```python
total_purchase_by_location = df.groupby('Location')['Purchase Amount (USD)'].sum()
print("Total Purchase Amount (USD) by Location:")
print(total_purchase_by_location)
```
